```python
##1. Multiplication table creator
def multiplier():
    while True:
        try:
            n = float(input("Enter a number to see its multiplication table: "))
            break
        except ValueError:
            print("Invalid input!")

    print(f"\nMultiplication Table for {n}:")
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")


multiplier()
```

```python
##2. Calculator
def adder(a, b):
    print(a+b)

def subtracter(a, b):
    print(a-b)

def multiplier(a, b):
    print(a*b)

def divider(a, b):
    if b == 0:
        print("Error: Cannot divide by zero!")
    else:
        print(a/b)

while True:
    print("Please enter two numbers: ")

    try:
        n1 = float(input())
    except ValueError:
        print("Invalid input!")
        continue

    try:
        n2 = float(input())
    except ValueError:
        print("Invalid input!")
        continue

    break

operation = str(input("Enter an operation to perform on the numbers: +,-,*, or/ \n"))

if operation == "+":
    adder(n1, n2)
elif operation == "-":
    subtracter(n1, n2)
elif operation == "*":
    multiplier(n1, n2)
elif operation == "/":
    divider(n1, n2)
else:
    print("Invalid input!")
```
